# Advanced Prompting

A short, practical introduction to writing clearer prompts for AI tools.

These examples are written as prompts you can try in an AI assistant. They do not require an API key.

## 1. Zero-shot prompting

Ask the model to do a task without giving an example.

**Prompt**

> Explain overfitting in machine learning in three simple sentences. Include one everyday example.

**Why it helps:** A clear task and a limit on the answer make the response easier to use.

## 2. Few-shot prompting

Give the model a couple of examples so it can follow the pattern.

**Prompt**

> Classify each review as Positive or Negative.  
> "The delivery was quick and the product works well." → Positive  
> "The item arrived broken." → Negative  
> "Good quality for the price." →  

**Expected label:** Positive

**Why it helps:** Examples show the model the format and labels you expect.

## 3. Role and context

A role can help set the level and point of view for an answer. It does not guarantee expertise or correctness.

**Prompt**

> Act as a patient machine learning tutor. Explain the difference between classification and regression to a beginner. Use one example of each.

## 4. Ask for structured output

Specify the fields and format you need. This is useful when the response will be processed by code.

**Prompt**

> Return only a valid JSON object for this dataset summary:  
> dataset: customer churn  
> rows: 5000  
> target: churn  
> Include the keys `dataset`, `row_count`, and `target`.

**Check:** If you use the result in a program, parse the JSON and handle invalid output. A prompt alone cannot guarantee valid JSON.

## 5. Improve a vague prompt

A vague prompt leaves important details open.

**Before**

> Explain decision trees.

**After**

> Explain how a decision tree makes a classification. Use a small customer-churn example, define the terms *root*, *split*, and *leaf*, and keep the explanation under 150 words.

**Takeaway:** Add the task, context, audience, and useful constraints.

## 6. Try it yourself

Choose a machine learning task and write a prompt for it. Then revise the prompt by adding missing context or a clear output format.

A simple checklist:

- Is the task specific?
- Did I provide the necessary context?
- Is the expected output clear?
- Can I check whether the answer is correct?

## Key point

Good prompting is an iterative process: write a prompt, review the response, and refine the instructions. Always verify important facts and outputs.
